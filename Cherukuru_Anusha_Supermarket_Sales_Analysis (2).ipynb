Supermarkets generate large amounts of transaction data every day. Simply storing this data does not provide useful information unless it is analyzed.

This project analyzes supermarket sales data to answer questions such as:

- Which products generate the highest sales?
- Which branch and city perform best?
- Which product category contributes the most revenue?
- Which payment method is most commonly used?
- How do Member and Normal customers differ in spending?
- How does sales quantity vary across products and categories?

- Are there relationships between quantity, price, sales, and ratings?

The final goal is to convert raw transaction data into understandable business insights.

## 4. Load the Dataset

```python
# Load the Excel dataset
file_path = "SUPER MARKET DATA.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="supermarket_sales_500_rows"
)

print("Dataset loaded successfully.")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Display the first five records
df.head()
```

```python
# Display the last five records
df.tail()
```

## 5. Initial Data Understanding

```python
# Dataset dimensions
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
for column in df.columns:
    print("-", column)
```

```python
# Data types of all columns
df.info()
```

```python
# Statistical summary of numeric columns
df.describe()
```

## 6. Data Cleaning

```python
# Remove unnecessary spaces from column names
df.columns = [column.strip() for column in df.columns]

# Convert Date to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Convert numeric columns to numeric data types
numeric_columns = ["Quantity", "Unit Price", "Rating", "Sales"]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print("Data types after cleaning:")
print(df.dtypes)
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Check unique values for important categorical columns
categorical_columns = [
    "Branch", "City", "Customer Type",
    "Gender", "Product", "Category", "Payment"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

## 7. Sales Calculation and Validation

```python
# Sales should equal Quantity × Unit Price
df["Calculated Sales"] = df["Quantity"] * df["Unit Price"]

# Compare calculated sales with the provided Sales column
df["Sales Difference"] = (
    df["Sales"] - df["Calculated Sales"]
).abs()

print("Maximum difference:", df["Sales Difference"].max())

if df["Sales Difference"].max() <= 0.01:
    print("Sales validation successful: Quantity × Unit Price matches Sales.")
else:
    print("Some sales values need further checking.")
```

## 8. Overall Sales Performance

```python
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
median_sales = df["Sales"].median()
total_quantity = df["Quantity"].sum()

print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Average Transaction Value: ₹{average_sales:,.2f}")
print(f"Median Transaction Value: ₹{median_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
```

## 9. Product-wise Sales Analysis

```python
product_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

product_sales
```

```python
# Highest and lowest sales products
highest_product = product_sales.idxmax()
lowest_product = product_sales.idxmin()

print(f"Highest-sales product: {highest_product}")
print(f"Sales: ₹{product_sales.max():,.2f}")

print(f"\nLowest-sales product: {lowest_product}")
print(f"Sales: ₹{product_sales.min():,.2f}")
```

```python
# Product quantity analysis
product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Top products by quantity sold:")
print(product_quantity.head(10))
```

```python
# Visualize top 10 products by sales
product_sales.head(10).sort_values().plot(
    kind="barh",
    figsize=(9, 5)
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

## 10. Branch-wise Sales Analysis

```python
branch_sales = (
    df.groupby(["Branch", "City"])["Sales"]
    .sum()
    .sort_values(ascending=False)
)

branch_sales
```

```python
# Branch transaction count and average sales
branch_summary = df.groupby("Branch").agg(
    Transactions=("Invoice ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values("Total_Sales", ascending=False)

branch_summary
```

```python
branch_summary["Total_Sales"].plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Total Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 11. City-wise Analysis

```python
city_summary = df.groupby("City").agg(
    Transactions=("Invoice ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values("Total_Sales", ascending=False)

city_summary
```

```python
city_summary["Total_Sales"].plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Sales by City")
plt.xlabel("City")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 12. Category-wise Analysis

```python
category_summary = df.groupby("Category").agg(
    Transactions=("Invoice ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
).sort_values("Total_Sales", ascending=False)

category_summary
```

```python
# Highest-selling category
highest_category = category_summary.index[0]

print("Highest-sales category:", highest_category)
print(
    f"Sales: ₹{category_summary.loc[highest_category, 'Total_Sales']:,.2f}"
)
```

```python
category_summary["Total_Sales"].sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Sales by Product Category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```

## 13. Payment Method Analysis

```python
payment_counts = df["Payment"].value_counts()

print("Payment method transaction counts:")
print(payment_counts)

print(
    f"\nMost-used payment method: "
    f"{payment_counts.index[0]} "
    f"({payment_counts.iloc[0]} transactions)"
)
```

```python
payment_sales = (
    df.groupby("Payment")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

payment_sales
```

```python
payment_counts.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Payment Method Usage")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 14. Customer Type Analysis

```python
customer_summary = df.groupby("Customer Type").agg(
    Transactions=("Invoice ID", "count"),
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Quantity_Sold=("Quantity", "sum")
)

customer_summary
```

## 19. Monthly Sales Analysis

## 22. Correlation Analysis

```python
correlation_columns = [
    "Quantity",
    "Unit Price",
    "Rating",
    "Sales"
]

correlation_matrix = df[correlation_columns].corr()

correlation_matrix
```
